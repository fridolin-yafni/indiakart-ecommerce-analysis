```python
cust = pd.read_csv("customers_clean.csv", dtype={"phone": str, "pincode": str})
print(cust.shape)    # expect (10000, 17)
cust.dtypes
```

```python
cust["email"] = cust["email"].str.lower().str.strip()

is_repeat = cust["email"].duplicated(keep="first")
local  = cust.loc[is_repeat, "email"].str.split("@").str[0]
domain = cust.loc[is_repeat, "email"].str.split("@").str[1]
cust.loc[is_repeat, "email"] = local + "+" + cust.loc[is_repeat, "customer_id"].str.lower() + "@" + domain

print(is_repeat.sum())                    
print(cust["email"].duplicated().sum())   
print(cust.loc[is_repeat, ["customer_id", "email"]])
```

```python
for col in ["registration_date", "last_login_date"]:
    cust[col] = pd.to_datetime(cust[col], format="%Y-%m-%d", errors="coerce")
print(cust[["registration_date", "last_login_date"]].isnull().sum())   

PERIOD_START, PERIOD_END = pd.Timestamp("2023-06-24"), pd.Timestamp("2025-06-23")

cust["registered_before_window"] = cust["registration_date"] < PERIOD_START
cust["login_after_window"]       = cust["last_login_date"] > PERIOD_END
print(cust[["registered_before_window", "login_after_window"]].sum())   

# Cap the 6 impossible logins at the end of the data period
cust.loc[cust["login_after_window"], "last_login_date"] = PERIOD_END
print((cust["last_login_date"] < cust["registration_date"]).sum())    
```

```python
orders_tmp = pd.read_csv("orders_clean.csv", usecols=["customer_id", "order_date"])
orders_tmp["order_date"] = pd.to_datetime(orders_tmp["order_date"], format="%Y-%m-%d", errors="coerce")
print(orders_tmp["order_date"].isnull().sum())        

last_order = orders_tmp.groupby("customer_id")["order_date"].max().rename("last_order_date")
cust = cust.merge(last_order, on="customer_id", how="left")
print(cust.shape)                                      
print(cust["last_order_date"].isnull().sum())        

def derive_segment(r):
    if r["total_orders"] == 0:
        return "Never Purchased"
    if (PERIOD_END - r["last_order_date"]).days > 180:
        return "Inactive"
    if r["total_orders"] >= 10:
        return "Premium"
    if r["total_orders"] >= 5:
        return "Regular"
    return "New/Low"

cust["segment_derived"] = cust.apply(derive_segment, axis=1)
print(cust["segment_derived"].value_counts())
```

```python
print((cust["last_order_date"] < cust["registration_date"]).sum())
print(cust.shape)    
```

```python
for col in ["registration_date", "last_login_date", "last_order_date"]:
    cust[col] = cust[col].dt.strftime("%Y-%m-%d")

cust.to_csv("customers_clean_v2.csv", index=False)

check = pd.read_csv("customers_clean_v2.csv")
print(check.shape, check["email"].duplicated().sum())    
```

```python
files = {
    "customers":   "customers_clean_v2.csv",
    "products":    "products_clean.csv",
    "orders":      "orders_clean.csv",
    "order_items": "order_items_clean.csv",
    "payments":    "payments_clean.csv",
    "returns":     "returns_clean.csv",
    "inventory":   "inventory_clean.csv",
    "suppliers":   "suppliers_clean.csv",
}
t = {name: pd.read_csv(f, dtype=str) for name, f in files.items()}

# 1. Row counts
for name, df in t.items():
    print(name, len(df))

# 2. Primary keys are unique
pks = {"customers":"customer_id","products":"product_id","orders":"order_id","order_items":"item_id",
       "payments":"payment_id","returns":"return_id","inventory":"inventory_id","suppliers":"supplier_id"}
for name, pk in pks.items():
    print("dup PK", name, t[name][pk].duplicated().sum())          

# 3. Foreign keys have no orphans
fks = [("orders","customer_id","customers","customer_id"),
       ("order_items","order_id","orders","order_id"),
       ("order_items","product_id","products","product_id"),
       ("payments","order_id","orders","order_id"),
       ("payments","customer_id","customers","customer_id"),
       ("returns","order_id","orders","order_id"),
       ("returns","customer_id","customers","customer_id"),
       ("inventory","product_id","products","product_id"),
       ("products","supplier_id","suppliers","supplier_id")]
for child, col, parent, pcol in fks:
    n = (~t[child][col].isin(t[parent][pcol])).sum()
    print(f"{child}.{col} -> {parent}.{pcol}: {n} orphans")        

# 4. Every date column uses YYYY-MM-DD (empty values are ignored)
date_cols = {"customers":["registration_date","last_login_date"], "products":["launch_date"],
             "orders":["order_date","delivered_date"], "payments":["payment_date","refund_date"],
             "returns":["return_date","refund_date"], "inventory":["last_restocked_date"],
             "suppliers":["created_date"]}
for name, cols in date_cols.items():
    for c in cols:
        s = t[name][c].dropna()
        bad = (~s.str.match(r"^\d{4}-\d{2}-\d{2}$")).sum()
        print(f"{name}.{c}: {bad} bad formats")                      

# 5. Email uniqueness for customers (the schema has a UNIQUE constraint)
print("customer email duplicates:", t["customers"]["email"].duplicated().sum())   
```
